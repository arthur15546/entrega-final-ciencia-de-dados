## 5 hoteis tem restaurantes

```python
df_amenities = (
    df_hotels[["hotel_id", "name", "amenities"]]
    .explode("amenities", ignore_index=True)
    .rename(columns={"amenities": "amenity"})
)
df_amenities["amenity"] = df_amenities["amenity"].str.lower().str.strip()

tem_restaurante = df_amenities["amenity"].str.contains(
    "restaurant",
    case=False,
    na=False
)

df_amenities.loc[
    tem_restaurante, "hotel_id"
].nunique()
```

## colunas interligadas
**hotle_id** liga todos arquivos de dados menos *reviews_scenarios.jsonl* 

**scenario** *liga reviews.jsonl* e *reviews_scenarios.jsonl*
